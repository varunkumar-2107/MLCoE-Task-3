```python
import pandas as pd
```

```python
df = pd.read_csv("EDA_done.csv")
df.drop(columns=['Unnamed: 0'], axis=0 , inplace=True)
```

```python
obj_cols = df.select_dtypes(include='object')
for col in obj_cols:
    print(col,"==>\n")
    print(df[col].unique())
    print("\n\n")
```

```python
df.drop(columns=['accident_id'] , axis=0 , inplace=True)
```

```python
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder(categories=[['No', 'Yes'],['No', 'Yes'],['No', 'Yes']],dtype=int)
df[['alcohol_involved','median_present','traffic_signal_present']] = oe.fit_transform(df[['alcohol_involved','median_present','traffic_signal_present']])
df[['hospitalization_required','police_called','ambulance_called']] = oe.fit_transform(df[['hospitalization_required','police_called','ambulance_called']])
df[['emergency_services_called','fatigue_involved','mobile_phone_used']] = oe.fit_transform(df[['emergency_services_called','fatigue_involved','mobile_phone_used']])


oe1 = OrdinalEncoder(categories=[['No', 'Yes'],['Streetlight Absent','Streetlight Present'],['Poor','Average','Good'],['Minor','Moderate','Severe','Fatal']],dtype=int)
df[['distraction_involved','road_lighting','vehicle_condition','accident_severity']] = oe1.fit_transform(df[['distraction_involved','road_lighting','vehicle_condition','accident_severity']])
```

```python
ohe_col = ['state','road_condition', 'road_type', 'area_type','license_status','driver_gender','weather_condition','junction_type','vehicle_type','primary_cause','helmet_seatbelt_used']
```

```python
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(dtype=int)
encoded = ohe.fit_transform(df[ohe_col])

encoded_df = pd.DataFrame(encoded.toarray(), columns=ohe.get_feature_names_out(ohe_col))

df = pd.concat([df, encoded_df], axis=1)
```

```python
df.drop(columns=ohe_col , axis=0 , inplace=True)
```
